```python
spark
df = spark.table("workspace.default.movie")
df.show(5)
```

```python
df.columns
```

```python
df.count()
```

```python
display(df.summary())
```

```python
df_trimmed = df.select("title","industry")
df_trimmed.show(3, truncate=True)
```

```python
from pyspark.sql.functions import col
df_filtered = df.filter((col("release_year") >= 2000) & (col("release_year") <= 2010))
df_filtered.show(5)
df_filtered.count()
```

```python
df_marvel = df.filter((col("studio") == "Marvel Studios") )
df_marvel.show()
```

```python
unique_industries = df.select("industry").distinct()
display(unique_industries)
```

```python
df_profit = df.withColumn("profit",col("revenue")-col("budget"))
display(df_profit)
```

```python
df_profit.withColumnRenamed("revenue","total_revenue")
df_profit.printSchema()
```
